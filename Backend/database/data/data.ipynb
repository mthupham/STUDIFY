# Thống kê số lượng câu hỏi

```python
import json
```

```python
file_path = "questionbank.json"

try:
    with open(file_path, "r", encoding="utf-8") as f:
        data = json.load(f)

    for lesson in data:
        topic_id = lesson.get("lesson_id", "N/A")
        topic_name = lesson.get("topic", "N/A")
        print(f"{topic_id}: {topic_name}")

        type_counts = {}

        for q in lesson.get("reading_questions", []):
            q_type = q.get("type", "unknown")
            type_counts[q_type] = type_counts.get(q_type, 0) + 1

        for q in lesson.get("writing_questions", []):
            q_type = q.get("type", "unknown")
            type_counts[q_type] = type_counts.get(q_type, 0) + 1

        total_questions = 0
        for q_type, count in type_counts.items():
            print(f"   - Dạng '{q_type}': {count} câu")
            total_questions += count

        print(f"Tổng số câu: {total_questions} câu\n" + "-" * 40)

except FileNotFoundError:
    print(f"Không tìm thấy file: {file_path}.")
except json.JSONDecodeError:
    print(" File JSON bị lỗi định dạng cấu trúc.")
```
